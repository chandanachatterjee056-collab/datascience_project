```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import skew, kurtosis, mode, pearsonr

# Read the data
df = pd.read_csv('student_grades.csv')

# Set student_id as index
df.set_index('student_id', inplace=True)

# List of subjects
subjects = ['math', 'science', 'cs', 'band', 'english', 'history', 'spanish', 'physed']

# Function to compute descriptive stats for a subject
def compute_stats(subject):
    data = df[subject]
    desc = data.describe()
    mean = desc['mean']
    median = data.median()
    mode_result = mode(data)
    mode_array = np.atleast_1d(mode_result.mode)
    count_array = np.atleast_1d(mode_result.count)
    mode_val = mode_array[0] if mode_array.size > 0 else np.nan
    std = desc['std']
    variance = data.var()
    min_val = desc['min']
    max_val = desc['max']
    range_val = max_val - min_val
    q1 = desc['25%']
    q3 = desc['75%']
    iqr = q3 - q1
    skewness = skew(data)
    kurt = kurtosis(data)

    # Print measurements
    print(f"\nDescriptive Statistics for {subject}:")
    print(f"1. Mean: {mean}")
    print(f"2. Median: {median}")
    print(f"3. Mode: {mode_val}")
    print(f"4. Standard Deviation: {std}")
    print(f"5. Variance: {variance}")
    print(f"6. Minimum: {min_val}")
    print(f"7. Maximum: {max_val}")
    print(f"8. Range: {range_val}")
    print(f"9. Q1 (25th Percentile): {q1}")
    print(f"10. Q3 (75th Percentile): {q3}")
    print(f"11. IQR: {iqr}")
    print(f"12. Skewness: {skewness}")
    print(f"13. Kurtosis: {kurt}")

    # Additional measurements
    print(f"14. Sum of grades: {data.sum()}")
    print(f"15. Count of grades: {data.count()}")
    print(f"16. SEM (Standard Error of Mean): {stats.sem(data)}")
    print(f"17. Coefficient of Variation: {std / mean * 100}%")
    print(f"18. 10th Percentile: {data.quantile(0.1)}")
    print(f"19. 90th Percentile: {data.quantile(0.9)}")
    print(f"20. Number of unique grades: {data.nunique()}")
    most_frequent_grade_count = count_array[0] if count_array.size > 0 else np.nan
    print(f"21. Most frequent grade count: {most_frequent_grade_count}")
    print(f"22. Proportion above 60: {(data > 60).mean() * 100}%")
    print(f"23. Proportion below 50: {(data < 50).mean() * 100}%")
    print(f"24. Median Absolute Deviation: {stats.median_abs_deviation(data)}")
    print(f"25. Trimmed Mean (10%): {stats.trim_mean(data, 0.1)}")

    return mean, median, mode_val, std, variance, min_val, max_val, range_val, q1, q3, iqr, skewness, kurt

# List to collect stats for all subjects
all_stats = {}

# Compute stats for each subject (that's 8 subjects * 25 measurements = 200, but we'll cap and add more)
for subject in subjects:
    all_stats[subject] = compute_stats(subject)

# Overall analysis
print("\nOverall Analysis Across All Subjects:")

# 26. Average grade across all subjects
overall_mean = df.mean(numeric_only=True).mean()
print(f"26. Overall Mean Grade: {overall_mean}")

# 27. Overall median
overall_median = df.stack().median()
print(f"27. Overall Median Grade: {overall_median}")

# 28. Overall mode
overall_mode_result = mode(df.stack())
overall_mode = np.atleast_1d(overall_mode_result.mode)[0] if overall_mode_result.mode.size > 0 else np.nan
print(f"28. Overall Mode: {overall_mode}")

# 29. Overall std
overall_std = df.stack().std()
print(f"29. Overall Standard Deviation: {overall_std}")

# 30. Subject with highest mean
highest_mean_subject = df.mean(numeric_only=True).idxmax()
print(f"30. Subject with Highest Mean: {highest_mean_subject} ({df.mean(numeric_only=True).max()})")

# 31. Subject with lowest mean
lowest_mean_subject = df.mean(numeric_only=True).idxmin()
print(f"31. Subject with Lowest Mean: {lowest_mean_subject} ({df.mean(numeric_only=True).min()})")

# 32. Student with highest average
df['average'] = df[subjects].mean(axis=1)
top_student = df['average'].idxmax()
print(f"32. Top Student ID: {top_student} with average {df['average'].max()}")

# 33. Student with lowest average
bottom_student = df['average'].idxmin()
print(f"33. Bottom Student ID: {bottom_student} with average {df['average'].min()}")

# 34. Correlation matrix
corr_matrix = df[subjects].corr() # Only subjects for correlation matrix
print("34. Correlation Matrix:")
print(corr_matrix)

# Pairwise correlations (for each pair, but count as one measurement with subparts)
pairs = [(subjects[i], subjects[j]) for i in range(len(subjects)) for j in range(i+1, len(subjects))]
for idx, (sub1, sub2) in enumerate(pairs, 35):
    corr, p = pearsonr(df[sub1], df[sub2])
    print(f"{idx}. Correlation between {sub1} and {sub2}: {corr} (p-value: {p})")

# Now, that's 34 + len(pairs) = 34 + 28 = 62 measurements so far (exceeding 50)

# Graphs: At least one graph per measurement type or per subject, but to reach related graphs for measurements

# 1-8: Histograms for each subject
for idx, subject in enumerate(subjects, 1):
    plt.figure()
    df[subject].hist(bins=20)
    plt.title(f"Histogram for {subject}")
    plt.xlabel("Grade")
    plt.ylabel("Frequency")
    # plt.savefig(f"histogram_{subject}.png")
    print(f"Graph {idx}: Histogram for {subject}")
    plt.show()

# 9: Boxplots for all subjects
plt.figure()
df.boxplot(column=subjects)
plt.title("Boxplots for All Subjects")
# plt.savefig("boxplots_all.png")
print("Graph 9: Boxplots for All Subjects")
plt.show()

# 10-16: Individual boxplots for each subject (starts from 10, total 8 subjects)
for idx, subject in enumerate(subjects, 10):
    plt.figure()
    df.boxplot(column=subject)
    plt.title(f"Boxplot for {subject}")
    # plt.savefig(f"boxplot_{subject}.png")
    print(f"Graph {idx}: Boxplot for {subject}")
    plt.show()

# 17: Correlation heatmap
plt.figure()
plt.imshow(corr_matrix, cmap='coolwarm')
plt.colorbar()
plt.xticks(range(len(subjects)), subjects, rotation=90)
plt.yticks(range(len(subjects)), subjects)
plt.title("Correlation Heatmap")
# plt.savefig("correlation_heatmap.png")
print("Graph 17: Correlation Heatmap")
plt.show()

# 18-25: Scatter plots for each pair (28 pairs, but let's do top 8 correlated pairs for brevity)
# Recalculate corr_matrix to ensure 'average' is not included if not desired for subject correlations
corr_matrix_subjects = df[subjects].corr()
sorted_pairs = sorted(pairs, key=lambda p: abs(corr_matrix_subjects.loc[p[0], p[1]]), reverse=True)[:8]
for idx, (sub1, sub2) in enumerate(sorted_pairs, 18):
    plt.figure()
    plt.scatter(df[sub1], df[sub2])
    plt.title(f"Scatter Plot: {sub1} vs {sub2}")
    plt.xlabel(sub1)
    plt.ylabel(sub2)
    # plt.savefig(f"scatter_{sub1}_vs_{sub2}.png")
    print(f"Graph {idx}: Scatter Plot {sub1} vs {sub2}")
    plt.show()

# 26: Bar plot of mean grades per subject
plt.figure()
df[subjects].mean().plot(kind='bar')
plt.title("Mean Grades per Subject")
plt.ylabel("Mean Grade")
# plt.savefig("mean_grades_bar.png")
print("Graph 26: Mean Grades Bar Plot")
plt.show()

# Additional graphs
# 28: Cumulative distribution for overall grades
plt.figure()
sorted_grades = np.sort(df[subjects].stack()) # Only subjects for stacking
plt.plot(sorted_grades, np.arange(len(sorted_grades)) / float(len(sorted_grades)))
plt.title("Cumulative Distribution of All Grades")
plt.xlabel("Grade")
plt.ylabel("Cumulative Proportion")
# plt.savefig("cumulative_dist.png")
print("Graph 28: Cumulative Distribution")
plt.show()

# 29: Pie chart of pass/fail overall
pass_count = (df[subjects] > 60).sum().sum()
fail_count = (df[subjects].size) - pass_count
plt.figure()
plt.pie([pass_count, fail_count], labels=['Pass', 'Fail'], autopct='%1.1f%%')
plt.title("Overall Pass/Fail Proportion")
# plt.savefig("pass_fail_pie.png")
print("Graph 29: Pass/Fail Pie Chart")
plt.show()

# 30: Line plot of top 10 student averages
top10 = df['average'].sort_values(ascending=False)[:10]
plt.figure()
top10.plot(kind='line')
plt.title("Top 10 Student Averages")
plt.ylabel("Average Grade")
# plt.savefig("top10_averages.png")
print("Graph 30: Top 10 Averages Line Plot")
plt.show()

# For example, QQ plots for normality check per subject
for idx, subject in enumerate(subjects, 31):
    plt.figure()
    stats.probplot(df[subject], dist="norm", plot=plt)
    plt.title(f"QQ Plot for {subject}")
    # plt.savefig(f"qq_{subject}.png")
    print(f"Graph {idx}: QQ Plot for {subject}")
    plt.show()

# Kernel density estimates
for idx, subject in enumerate(subjects, 39):
    plt.figure()
    df[subject].plot(kind='kde')
    plt.title(f"KDE for {subject}")
    # plt.savefig(f"kde_{subject}.png")
    print(f"Graph {idx}: KDE for {subject}")
    plt.show()

# Bar plots for count above/below thresholds per subject
for idx, subject in enumerate(subjects, 47):
    counts = [(df[subject] > 60).sum(), (df[subject] <= 60).sum()]
    plt.figure()
    plt.bar(['Pass', 'Fail'], counts)
    plt.title(f"Pass/Fail Bar for {subject}")
    # plt.savefig(f"pass_fail_bar_{subject}.png")
    print(f"Graph {idx}: Pass/Fail Bar for {subject}")
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import skew, kurtosis, pearsonr

# Read the data
df = pd.read_csv('student_grades.csv')

# Set student_id as index (only if present)
if 'student_id' in df.columns:
    df.set_index('student_id', inplace=True)

# List of subjects (ensure these columns exist)
subjects = ['math', 'science', 'cs', 'band', 'english', 'history', 'spanish', 'physed']
subjects = [s for s in subjects if s in df.columns]

# Function to compute descriptive stats for a subject
def compute_stats(subject):
    data = df[subject].dropna()
    desc = data.describe()

    mean = desc['mean']
    median = data.median()

    # Mode using pandas (scipy.stats.mode is deprecated / changing behavior)
    mode_series = data.mode()
    mode_val = mode_series.iloc[0] if len(mode_series) > 0 else np.nan

    std = desc['std']
    variance = data.var()
    min_val = desc['min']
    max_val = desc['max']
    range_val = max_val - min_val
    q1 = desc['25%']
    q3 = desc['75%']
    iqr = q3 - q1
    skewness = skew(data, bias=False)
    kurt = kurtosis(data, bias=False)

    # Print measurements
    print(f"\nDescriptive Statistics for {subject}:")
    print(f"1. Mean: {mean}")
    print(f"2. Median: {median}")
    print(f"3. Mode: {mode_val}")
    print(f"4. Standard Deviation: {std}")
    print(f"5. Variance: {variance}")
    print(f"6. Minimum: {min_val}")
    print(f"7. Maximum: {max_val}")
    print(f"8. Range: {range_val}")
    print(f"9. Q1 (25th Percentile): {q1}")
    print(f"10. Q3 (75th Percentile): {q3}")
    print(f"11. IQR: {iqr}")
    print(f"12. Skewness: {skewness}")
    print(f"13. Kurtosis: {kurt}")

    # Additional measurements
    print(f"14. Sum of grades: {data.sum()}")
    print(f"15. Count of grades: {data.count()}")
    print(f"16. SEM (Standard Error of Mean): {stats.sem(data, nan_policy='omit')}")
    print(f"17. Coefficient of Variation: {std / mean * 100 if mean != 0 else np.nan}%")
    print(f"18. 10th Percentile: {data.quantile(0.1)}")
    print(f"19. 90th Percentile: {data.quantile(0.9)}")
    print(f"20. Number of unique grades: {data.nunique()}")

    # Mode count using pandas value_counts
    value_counts = data.value_counts()
    most_freq_count = value_counts.iloc[0] if len(value_counts) > 0 else np.nan
    print(f"21. Most frequent grade count: {most_freq_count}")

    print(f"22. Proportion above 60: {(data > 60).mean() * 100}%")
    print(f"23. Proportion below 50: {(data < 50).mean() * 100}%")
    print(f"24. Median Absolute Deviation: {stats.median_abs_deviation(data, nan_policy='omit')}")
    print(f"25. Trimmed Mean (10%): {stats.trim_mean(data, 0.1)}")

    return (
        mean, median, mode_val, std, variance, min_val, max_val, range_val,
        q1, q3, iqr, skewness, kurt
    )

# Dictionary to collect stats for all subjects
all_stats = {}

# Compute stats for each subject
for subject in subjects:
    all_stats[subject] = compute_stats(subject)

# Overall analysis
print("\nOverall Analysis Across All Subjects:")

# Use only subject columns for overall stats
df_subjects = df[subjects]

# 26. Average grade across all subjects
overall_mean = df_subjects.mean().mean()
print(f"26. Overall Mean Grade: {overall_mean}")

# 27. Overall median
overall_median = df_subjects.stack().median()
print(f"27. Overall Median Grade: {overall_median}")

# 28. Overall mode (using pandas mode)
overall_mode_series = df_subjects.stack().mode()
overall_mode = overall_mode_series.iloc[0] if len(overall_mode_series) > 0 else np.nan
print(f"28. Overall Mode: {overall_mode}")

# 29. Overall std
overall_std = df_subjects.stack().std()
print(f"29. Overall Standard Deviation: {overall_std}")

# 30. Subject with highest mean
subject_means = df_subjects.mean()
highest_mean_subject = subject_means.idxmax()
print(f"30. Subject with Highest Mean: {highest_mean_subject} ({subject_means.max()})")

# 31. Subject with lowest mean
lowest_mean_subject = subject_means.idxmin()
print(f"31. Subject with Lowest Mean: {lowest_mean_subject} ({subject_means.min()})")

# 32. Student with highest average
df['average'] = df_subjects.mean(axis=1)
top_student = df['average'].idxmax()
print(f"32. Top Student ID: {top_student} with average {df['average'].max()}")

# 33. Student with lowest average
bottom_student = df['average'].idxmin()
print(f"33. Bottom Student ID: {bottom_student} with average {df['average'].min()}")

# 34. Correlation matrix (only subjects)
corr_matrix = df_subjects.corr()
print("34. Correlation Matrix:")
print(corr_matrix)

# Pairwise correlations
pairs = [
    (subjects[i], subjects[j])
    for i in range(len(subjects))
    for j in range(i + 1, len(subjects))
]
for idx, (sub1, sub2) in enumerate(pairs, 35):
    corr, p = pearsonr(df[sub1], df[sub2])
    print(f"{idx}. Correlation between {sub1} and {sub2}: {corr} (p-value: {p})")

# ==========================
# GRAPHS
# ==========================

# 1–8: Histograms for each subject
for idx, subject in enumerate(subjects, 1):
    plt.figure()
    df[subject].hist(bins=20)
    plt.title(f"Histogram for {subject}")
    plt.xlabel("Grade")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.savefig(f"histogram_{subject}.png")
    plt.close()
    print(f"Graph {idx}: Histogram for {subject} saved as histogram_{subject}.png")

# 9: Boxplots for all subjects together
plt.figure()
df_subjects.boxplot(column=subjects)
plt.title("Boxplots for All Subjects")
plt.ylabel("Grade")
plt.tight_layout()
plt.savefig("boxplots_all.png")
plt.close()
print("Graph 9: Boxplots for All Subjects saved as boxplots_all.png")

# 10–17: Individual boxplots
for idx, subject in enumerate(subjects, 10):
    plt.figure()
    df.boxplot(column=subject)
    plt.title(f"Boxplot for {subject}")
    plt.ylabel("Grade")
    plt.tight_layout()
    plt.savefig(f"boxplot_{subject}.png")
    plt.close()
    print(f"Graph {idx}: Boxplot for {subject} saved as boxplot_{subject}.png")

# 18: Correlation heatmap
plt.figure()
plt.imshow(corr_matrix, cmap='coolwarm', aspect='auto')
plt.colorbar()
plt.xticks(range(len(subjects)), subjects, rotation=90)
plt.yticks(range(len(subjects)), subjects)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("correlation_heatmap.png")
plt.close()
print("Graph 18: Correlation Heatmap saved as correlation_heatmap.png")

# 19–26: Scatter plots for top 8 correlated pairs
sorted_pairs = sorted(
    pairs,
    key=lambda p: abs(corr_matrix.loc[p[0], p[1]]),
    reverse=True
)[:8]

for idx, (sub1, sub2) in enumerate(sorted_pairs, 19):
    plt.figure()
    plt.scatter(df[sub1], df[sub2])
    plt.title(f"Scatter Plot: {sub1} vs {sub2}")
    plt.xlabel(sub1)
    plt.ylabel(sub2)
    plt.tight_layout()
    plt.savefig(f"scatter_{sub1}_vs_{sub2}.png")
    plt.close()
    print(f"Graph {idx}: Scatter Plot {sub1} vs {sub2} saved as scatter_{sub1}_vs_{sub2}.png")

# 27: Bar plot of mean grades per subject
plt.figure()
subject_means.plot(kind='bar')
plt.title("Mean Grades per Subject")
plt.ylabel("Mean Grade")
plt.tight_layout()
plt.savefig("mean_grades_bar.png")
plt.close()
print("Graph 27: Mean Grades Bar Plot saved as mean_grades_bar.png")

# 28: Cumulative distribution for overall grades
plt.figure()
sorted_grades = np.sort(df_subjects.stack())
plt.plot(sorted_grades, np.arange(len(sorted_grades)) / float(len(sorted_grades)))
plt.title("Cumulative Distribution of All Grades")
plt.xlabel("Grade")
plt.ylabel("Cumulative Proportion")
plt.tight_layout()
plt.savefig("cumulative_dist.png")
plt.close()
print("Graph 28: Cumulative Distribution saved as cumulative_dist.png")

# 29: Pie chart of pass/fail overall
all_grades = df_subjects
pass_count = (all_grades > 60).sum().sum()
fail_count = all_grades.size - pass_count
plt.figure()
plt.pie([pass_count, fail_count], labels=['Pass', 'Fail'], autopct='%1.1f%%')
plt.title("Overall Pass/Fail Proportion")
plt.tight_layout()
plt.savefig("pass_fail_pie.png")
plt.close()
print("Graph 29: Pass/Fail Pie Chart saved as pass_fail_pie.png")

# 30: Line plot of top 10 student averages
top10 = df['average'].sort_values(ascending=False).head(10)
plt.figure()
top10.plot(kind='line', marker='o')
plt.title("Top 10 Student Averages")
plt.xlabel("Student ID")
plt.ylabel("Average Grade")
plt.tight_layout()
plt.savefig("top10_averages.png")
plt.close()
print("Graph 30: Top 10 Averages Line Plot saved as top10_averages.png")

# 31–38: QQ plots for each subject
for idx, subject in enumerate(subjects, 31):
    plt.figure()
    stats.probplot(df[subject].dropna(), dist="norm", plot=plt)
    plt.title(f"QQ Plot for {subject}")
    plt.tight_layout()
    plt.savefig(f"qq_{subject}.png")
    plt.close()
    print(f"Graph {idx}: QQ Plot for {subject} saved as qq_{subject}.png")

# 39–46: Kernel density estimates
for idx, subject in enumerate(subjects, 39):
    plt.figure()
    df[subject].dropna().plot(kind='kde')
    plt.title(f"KDE for {subject}")
    plt.xlabel("Grade")
    plt.tight_layout()
    plt.savefig(f"kde_{subject}.png")
    plt.close()
    print(f"Graph {idx}: KDE for {subject} saved as kde_{subject}.png")

# 47–54: Bar plots for count above/below threshold per subject
for idx, subject in enumerate(subjects, 47):
    subject_data = df[subject].dropna()
    counts = [(subject_data > 60).sum(), (subject_data <= 60).sum()]
    plt.figure()
    plt.bar(['Pass', 'Fail'], counts)
    plt.title(f"Pass/Fail Bar for {subject}")
    plt.ylabel("Count")
    plt.tight_layout()
    plt.savefig(f"pass_fail_bar_{subject}.png")
    plt.close()
    print(f"Graph {idx}: Pass/Fail Bar for {subject} saved as pass_fail_bar_{subject}.png")

print("\nAnalysis complete. Measurements and graphs generated.")
```
